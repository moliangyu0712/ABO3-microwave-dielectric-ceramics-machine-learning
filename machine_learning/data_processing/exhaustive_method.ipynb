# feature_combination.csv (Qf~features)

```python
#GetTargetName
a = pd.DataFrame(list(data))
goal = a.iloc[-1, :1][0]
# GetFeatureCombination
feature_combination = pd.DataFrame()
features = pd.DataFrame(data.columns)
features.columns = ['feature']
features = features.iloc[:-1,:] #DeleteIsOutputColumn,Need to adjust based on actual conditions
for i in range(1,len(features)+1):
    num_features = i
    # ObtainAllPossibleFeaturecombination patterns
    combinations = list(itertools.combinations(features['feature'], num_features))
    # OutputResult
    for c in combinations:
        s = '+'.join(c)
        name = goal+'~'+s
        feature_combination = feature_combination._append({'feature':name}, ignore_index=True)
```

```python
feature_combination.to_csv(f'{dizhi}feature_combination.csv',index = False)
```

# feature_combination.file

```python
feature_combination = pd.DataFrame()
features = pd.DataFrame(data.columns)
features.columns = ['feature']
features = features.iloc[:-1,:] #DeleteIsOutputColumn,Need to adjust based on actual conditions
for i in range(1,len(features)+1):
    num_features = i
    # ObtainAllPossibleFeaturecombination patterns
    combinations = list(itertools.combinations(features['feature'], num_features))
    # OutputResult
    for c in combinations:
        s = '+'.join(c)
        feature_combination = feature_combination._append({'feature':s}, ignore_index=True)
for i in feature_combination.values.tolist():
    result = pd.DataFrame()
    after_split = i[0].split("+")
    for j in after_split:
        result[j] = data[j]
    result['targ'] = data['targ']
    name = 'targ'+'~'+i[0]
    result.to_csv(f'{dizhi}feature_combination/{name}.csv', index=False)
```
