## 5cv_result_r2_te

```python
dire = "D:/MLY/ABO3/Machine learning process/code-6/S3-feature_selection/RF/5cv_result/"
```

```python
import math
r2_te = pd.DataFrame()
x = 1
for i in importance_combination.values.tolist():
    print(f"{x}.{i[0]}")
    x = x + 1
    for k in range(2, 8):
        r2_te_sum = pd.DataFrame(0 , index = [0, 1, 2, 3, 4, 5, 6, 7], 
                             columns = ['V1', 'V2', 'V3', 'V4', 'V5', 'V6', 'V7', 'V8', 'V9', 'V10'])
        fc_max_r2_te = -math.inf
        for j in range(1, 31):
            name = f'{j}_{i[0]}_{k}_r2_test.csv'
            file = pd.read_csv(os.path.join(dire, name))
            file = file.iloc[:,1:] #SelectdataRowandColumn
            r2_te_sum = r2_te_sum + file
        r2_te_sum = r2_te_sum/30 #Computeaverage
        r2_te_sum.index = [3, 6, 9, 12, 15, 18, 21, 24]
        r2_te_sum.columns = [100, 200, 300, 400, 500, 600, 700, 800, 900, 1000]
        r2_te_sum_find = r2_te_sum.max()
        if r2_te_sum_find.max() > fc_max_r2_te:
            fc_max_r2_te = r2_te_sum_find.max()
            mtry = k
        for index, row in r2_te_sum.iterrows():
            for column in r2_te_sum.columns:
                value = row[column]
                if value == fc_max_r2_te:
                    r2_te = r2_te._append({'features':i[0], 'best_hp_result':fc_max_r2_te, 
                                            'maxnode':index, 'ntree':column, 'mtry':mtry}, ignore_index = True)
```

```python
result = pd.DataFrame()
r2_te['best_mtry'] = r2_te.groupby('features')['best_hp_result'].transform('max')
for row in r2_te.values.tolist():
    feature, bhr, maxnode, ntree, mtry, bmtry = row
    if bmtry == bhr:
        result = result._append({'features':feature, 'best_hp_result':bhr, 'maxnode':maxnode, 'ntree':ntree, 'mtry':mtry}, ignore_index = True)
result.drop_duplicates(subset= ['best_hp_result'],keep='first',inplace=True)
```

```python
result.to_csv('D:/MLY/ABO3/Machine learning process/code-10/S3-feature_selection/RF/r2_te.csv', index = False)
```
