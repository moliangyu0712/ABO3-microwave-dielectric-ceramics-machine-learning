```python
import numpy as np
import pandas as pd

def calculate_r2(y_true, y_pred):
    """
    Based on R2 DefineFormulaManualCalculateCoefficient of determinationCount
    :param y_true: TrueActualValueColumnTable/CountGroup
    :param y_pred: PredictionValueColumnTable/CountGroup
    :return: R2 SplitCount
    """
    # 1. ConvertAs numpy CountGroup(ConvenientCalculate)
    y_true = np.array(y_true)
    y_pred = np.array(y_pred)
    
    # 2. CalculateTrueActualValueaverageValue
    y_mean = np.mean(y_true)
    
    # 3. TotalSquareand SST = sum( (TrueActualValue - TrueActualValueAverageValue)^2 )
    sst = np.sum((y_true - y_mean) ** 2)
    
    # 4. ResidualPoorSquareand SSE = sum( (TrueActualValue - PredictionValue)^2 )
    sse = np.sum((y_true - y_pred) ** 2)
    
    # 5. According toDefineCalculate R2
    r2 = 1 - (sse / sst)
    
    return r2

def calculate_rmse(y_true, y_pred):
    """
    ManualCalculateAverageError RMSE
    :param y_true: TrueActualValue
    :param y_pred: PredictionValue
    :return: RMSE
    """
    y_true = np.array(y_true)
    y_pred = np.array(y_pred)
    # AverageSquareError MSE
    mse = np.mean((y_true - y_pred) ** 2)
    # AverageError RMSE
    rmse = np.sqrt(mse)
    return rmse

def calculate_mae(y_true, y_pred):
    """
    ManualCalculateaverageAbsoluteCorrectError MAE
    Formula: MAE = average(|TrueActualValue - PredictionValue|)
    """
    y_true = np.array(y_true)
    y_pred = np.array(y_pred)
    mae = np.mean(np.abs(y_true - y_pred))
    return mae

# Readdata
data = pd.read_csv("D:/MLY/ABO3/Machine learning process/code-23/data/PredictionResultEvaluation_new.csv")

# ===================== KeyFixed: ConvertAsCountValueType =====================
y = pd.to_numeric(data["targ"], errors='coerce')  # ForceConvertNumber
y_pred = pd.to_numeric(data["train_mean"], errors='coerce')  # ForceConvertNumber
# CalculateThreeMetric
r2 = calculate_r2(y, y_pred)
rmse = calculate_rmse(y, y_pred)
mae = calculate_mae(y, y_pred)

# OutputResult
print(f'R²:    {r2:.4f}')
print(f'RMSE:  {rmse:.4f}')
print(f'MAE:   {mae:.4f}')
```
